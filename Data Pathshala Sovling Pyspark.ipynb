Task: return the department and the count of employees (employee_count) for departements that have 5 or more employees .Order the results by employee_count desc

```python
from pyspark.sql import functions as F
```

```python
data = [
    (1, "Alice", "Engineering", 95000),
    (2, "Bob", "Engineering", 88000),
    (3, "Carol", "Engineering", 91000),
    (4, "David", "Engineering", 103000),
    (5, "Elena", "Engineering", 79000),
    (6, "Frank", "Engineering", 85000),
    (7, "Grace", "Marketing", 72000),
    (8, "Hank", "Marketing", 68000),
    (9, "Iris", "Marketing", 75000),
    (10, "Jake", "Marketing", 70000),
    (11, "Karen", "HR", 65000),
    (12, "Leo", "HR", 62000),
    (13, "Mona", "Finance", 98000),
    (14, "Nina", "Finance", 105000),
    (15, "Oscar", "Sales", 73000)
]

# Column names
columns = ["id", "name", "department", "salary"]

# Create DataFrame
df = spark.createDataFrame(data, columns)

# df.show()

# select_df = df.groupBy('department').agg(F.count('id').alias('employee_count'))\
#     .filter(F.col('employee_count')>3)

elect_df = df.groupBy('department').count().alias('employee_count')\
    .filter(F.col('employee_count')>3)
display(select_df)
```

```python
distinct_depts = df.groupBy('department').agg(F.countDistinct('id').alias('unique_depts'))
display(distinct_depts)
```

```python
grouped_df = df.groupBy("department").agg(F.count("id").alias("employee_count"),
                                          F.sum("salary").alias('total_salary'),
                                          F.avg('salary').alias('average_salary'),
                                          F.max('salary').alias('maximum_salary'),
                                          F.min('salary').alias('minimum_salary'),
                                          F.stddev('salary').alias('stddev_salary'),
                                          F.median('salary').alias('median_salary'),
                                          F.variance('salary').alias('variance_salary'),
                                          F.skewness('salary').alias('skewness_salary'),
                                          F.kurtosis('salary').alias('kurtosis_salary'),
                                          F.mode('salary').alias('mode_salary'),
                                          F.percentile_approx('salary', [0.25, 0.5, 0.75]).alias('percentile_salary')
                                          )
display(grouped_df)
```

# Task 2:
Return phone_number and count_occurences for every phone number that appears more than once in the table ,Order the results by count_occurences Desc then phoneNumber asc

```python
# Contacts data
data = [
    (1, "Alice Johnson", "555-0191"),
    (2, "Bob Smith", "555-0142"),
    (3, "Carol White", "555-0191"),
    (4, "David Brown", "555-0178"),
    (5, "Eva Martinez", "555-0191"),
    (6, "Frank Davis", "555-0142"),
    (7, "Grace Lee", "555-0305"),
    (8, "Henry Wilson", "555-0267")
]

# Column names
columns = ["id", "name", "phone_number"]

# Create DataFrame
contacts_df = spark.createDataFrame(data, columns)

group_contact_df = contacts_df.groupBy('phone_number').agg(F.count('phone_number').alias('count_occurrences'))\
    .filter(F.col('count_occurrences')>1)\
    .orderBy('count_occurrences','phone_number',ascending=[False,True])
    

# Show DataFrame
display(group_contact_df)
```

# Task 3
Return department and active_count (count of active employees) for departments that have at least one active employee. Order results by active_count DESC, then department ASC to break ties

```python
data = [
    (1, "Alice Nguyen", "Engineering", "active"),
    (2, "Bob Martinez", "Engineering", "active"),
    (11,'Jhone Doe',"Engineering","active"),
    (3, "Carol Smith", "Engineering", "inactive"),
    (4, "David Lee", "Marketing", "active"),
    (5, "Eva Brown", "Marketing", "inactive"),
    (6, "Frank Wilson", "Marketing", "inactive"),
    (7, "Grace Kim", "Sales", "active"),
    (8, "Henry Chen", "Sales", "active"),
    (9, "Irene Park", "Sales", "active"),
    (10, "James Taylor", "HR", "inactive")
]

# Column names
columns = ["id", "name", "department", "status"]

# Create DataFrame
employees_df = spark.createDataFrame(data, columns)

grouped_df = employees_df.filter(F.col('status') == 'active').groupBy("department")\
    .agg(F.count('department').alias('active_count'))\
    .orderBy("active_count","department",ascending=[False,True])

# Show DataFrame
display(grouped_df)
```

#Task 4
Return name and email of all customers whose email address ends with '@gmail.com', ordered by name ASC

```python
data = [
    (1, "Alice Nguyen", "alice.nguyen@gmail.com", "San Francisco"),
    (2, "Bob Smith", "bob.smith@yahoo.com", "Austin"),
    (3, "Carol White", "carol.white@gmail.com", "Chicago"),
    (4, "David Lee", "david.lee@outlook.com", "Seattle"),
    (5, "Eva Martinez", "eva.martinez@gmail.com", "Miami"),
    (6, "Frank Davis", "frank.davis@hotmail.com", "Denver"),
    (7, "Grace Kim", "grace.kim@gmail.com", "New York")
]

# Column names
columns = ["id", "name", "email", "city"]

# Create DataFrame
customers_df = spark.createDataFrame(data, columns)

grouped_df = customers_df.select("name","email").filter(F.split(F.col('email'), '@')[1] == 'gmail.com')

# Show DataFrame
display(grouped_df)
```

#Task 5
Return name and rounded_price (price rounded to the nearest whole number ) orderde by roundedprice Asc then name asc

```python
data = [
    (1, "Bluetooth Speaker", 29.10),
    (2, "Phone Case", 9.50),
    (3, "Screen Protector", 4.49),
    (4, "Charging Cable", 14.99),
    (5, "Portable Charger", 24.75),
    (6, "Earbuds", 19.50)
]

# Column names
columns = ["id", "name", "price"]

# Create DataFrame
products_df = spark.createDataFrame(data, columns)
select_df = products_df.select("name",F.round("price",scale=0).alias('rounded_price')).orderBy(F.col('rounded_price'),"name")
# Show DataFrame
select_df.show(truncate=False)
```
