```python
emp_data = [
(1,'manish',26,20000,'india','IT'),
(2,'rahul',None,40000,'germany','engineering'),
(3,'pawan',12,60000,'india','sales'),
(4,'roshini',44,None,'uk','engineering'),
(5,'raushan',35,70000,'india','sales'),
(6,None,29,200000,'uk','IT'),
(7,'adam',37,65000,'us','IT'),
(8,'chris',16,40000,'us','sales'),
(None,None,None,None,None,None),
(7,'adam',37,65000,'us','IT')
]

emp_df = spark.createDataFrame(emp_data,['id','name','age','salary','country',"dept"])
emp_df.show()
emp_df.createOrReplaceTempView("empdata")
```

```python
emp_df.withColumn("age",when(col("age").isNull(), lit(18)).otherwise(col("age")))\
.withColumn("adult",when(col("age") > 18,"Yes").otherwise("No")).show()
```

```python
spark.sql("""
          select *,
          case
            when salary > 50000 then 'senior'
            when salary > 30000 then 'junior'
            else 'fresher'
          end as status
          from empdata
          
          """).show()
```

```python
leet_code_data = [
    (1, 'Will', None),
    (2, 'Jane', None),
    (3, 'Alex', 2),
    (4, 'Bill', None),
    (5, 'Zack', 1),
    (6, 'Mark', 2)
]
leet_code_df = spark.createDataFrame(leet_code_data, ['id', 'name', 'refer_id'])
leet_code_df.createOrReplaceTempView("leet_code")
```
