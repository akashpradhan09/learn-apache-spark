#Spark Application

```python
spark = SparkSession.builder.appName("asddd").getOrCreate()
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
flight_data_path = "/Volumes/spark_data/dev/spark_data_volume/flight-data/csv/2014-summary.csv"
```

##  Read which is wide transformation

```python
flight_df = spark.read.format("csv")\
    .option("header","true")\
    .option("inferSchema","true")\
    .load(flight_data_path)
```

```python
print(flight_df.rdd.getNumPartitions())  # Not supported in shared clusters, consider alternative methods.
```

## filter Narrow transformation

```python
flight_repartion_df =flight_df.repartition(3)
```

```python
us_flight_df = flight_df.where(col("DEST_COUNTRY_NAME")=="UnitedS tates")
us_ind_flight_df = us_flight_df.filter((col("ORIGIN_COUNTRY_NAME")=='India') |(col("ORIGIN_COUNTRY_NAME")=="Singapore"))
```

## Groupby wide dependency transformation 

```python
total_flight_ind_sing = us_ind_flight_df.groupBy("DEST_COUNTRY_NAME").sum("count")
```
