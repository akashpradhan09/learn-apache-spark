```python
from pyspark.sql.types import *
from pyspark.sql.functions  import*
```

```python
customer_data = [(1,'manish','patna',"30-05-2022"),
(2,'vikash','kolkata',"12-03-2023"),
(3,'nikita','delhi',"25-06-2023"),
(4,'rahul','ranchi',"24-03-2023"),
(5,'mahesh','jaipur',"22-03-2023"),
(6,'prantosh','kolkata',"18-10-2022"),
(7,'raman','patna',"30-12-2022"),
(8,'prakash','ranchi',"24-02-2023"),
(9,'ragini','kolkata',"03-03-2023"),
(10,'raushan','jaipur',"05-02-2023"),
(None,None,None,None,)]

customer_schema=['customer_id','customer_name','address','date_of_joining']


sales_data = [(1,22,10,"01-06-2022"),
(1,27,5,"03-02-2023"),
(2,5,3,"01-06-2023"),
(5,22,1,"22-03-2023"),
(7,22,4,"03-02-2023"),
(9,5,6,"03-03-2023"),
(2,1,12,"15-06-2023"),
(1,56,2,"25-06-2023"),
(5,12,5,"15-04-2023"),
(11,12,76,"12-03-2023"),
(None,None,None,None)]

sales_schema=['customer_id','product_id','quantity','date_of_purchase']


product_data = [(1, 'fanta',20),
(2, 'dew',22),
(5, 'sprite',40),
(7, 'redbull',100),
(12,'mazza',45),
(22,'coke',27),
(25,'limca',21),
(27,'pepsi',14),
(56,'sting',10)]

product_schema=['id','name','price']
```

```python
customer_df = spark.createDataFrame(customer_data,customer_schema)
customer_df.createOrReplaceTempView("customer_tbl")
sales_df = spark.createDataFrame(sales_data,sales_schema)
sales_df.createOrReplaceTempView("sales_tbl")
product_df = spark.createDataFrame(product_data,product_schema)
product_df.createOrReplaceTempView("product_tbl")
```

```python
customer_df.show()
sales_df.show()
product_df.show()
```

```python
inner_cust_df = customer_df.join(sales_df,sales_df.customer_id == customer_df.customer_id,"left_anti")
display(inner_cust_df)
```

```python
inner_cust_df = customer_df.join(sales_df,sales_df.customer_id == customer_df.customer_id,"inner")
display(inner_cust_df)
```

```python
outer_cust_df = customer_df.join(sales_df,sales_df.customer_id == customer_df.customer_id,"left")
display(outer_cust_df)
```

```python
inner_cust_df = customer_df.join(sales_df,(sales_df.customer_id == customer_df.customer_id,"inner")\
    .select(sales_df.product_id,customer_df.customer_id).sort("product_id",customer_df.customer_id)
```

```python
inner_cust_df.join(product_df,product_df.id == inner_cust_df.product_id,"inner").show()
```

```python
person_data = [
    (1, "Wang", "Allen"),
    (2, "Alice", "Bob")
]

person = spark.createDataFrame(person_data, ["personId", "lastName", "firstName"])

person.show()
```

```python
address_data = [
    (1, 2, "New York City", "New York"),
    (2, 3, "Leetcode", "California")
]

address = spark.createDataFrame(address_data, ["addressId", "personId", "city", "state"])

address.show()
```

```python
person.join(address,person.personId==address.personId,"left")\
    .select(person.firstName,person.lastName,address.city,address.state)\
        .show()
```
