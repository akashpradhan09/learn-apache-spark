```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
```

```python
flight_data_path = "/Volumes/spark_data/dev/spark_data_volume/flight-data/csv/2010-summary.csv"
```

```python
flight_df = spark.read.format("csv").option("header","true").option("inferSchema","true").load(flight_data_path)
```

```python
flight_df.count()
```

```python
flight_df.rdd.getNumPartitions()
```

```python
part_df = flight_df.repartition(4)
```

```python
part_df.rdd.getNumPartitions()
```

```python
part_groupby_df = part_df.withColumn("partitionId",spark_partition_id()).groupBy('partitionId').agg(count("*"))
part_groupby_df.show()
```

```python
partition_on_column = flight_df.repartition(300,"ORIGIN_COUNTRY_NAME")
partition_on_column.rdd.getNumPartitions()
```

```python
coalesce_flight_df = flight_df.repartition(8)
```

```python
colaesce_group_df = coalesce_flight_df.withColumn("partitionId",spark_partition_id()).groupBy("partitionId").agg(count('*'))

colaesce_group_df.show()
```

```python
three_coalesce_df = coalesce_flight_df.coalesce(3)
three_coalesce_df.rdd.getNumPartitions()
```

```python
three_coalesce_d1f = three_coalesce_df.coalesce(13)
three_coalesce_d1f.rdd.getNumPartitions()
```

```python
three_coalesce_df = three_coalesce_df.withColumn("partitionId",spark_partition_id()).groupBy('partitionId').agg(count('*'))

three_coalesce_df.show()
```
